```python
#2.Image Color Inspector
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
img = cv2.imread("passport_image.jfif")

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)


hue = hsv[:, :, 0]


img_small = cv2.resize(img, (300, 200))
gray_small = cv2.resize(gray, (300, 200))
hsv_small = cv2.resize(hsv, (300, 200))
hue_small = cv2.resize(hue, (300, 200))

top = np.hstack((img_small, cv2.cvtColor(gray_small, cv2.COLOR_GRAY2BGR)))
bottom = np.hstack((hsv_small, cv2.cvtColor(hue_small, cv2.COLOR_GRAY2BGR)))

result = np.vstack((top, bottom))

cv2_imshow(result)

print("Hue Channel:")
cv2_imshow(hue)
```

```python
#3.Profile Picture Editor
import cv2
from google.colab.patches import cv2_imshow
img = cv2.imread("passport_image.jfif")
h, w = img.shape[:2]

size = min(h, w)

x = (w - size) // 2
y = (h - size) // 2

square = img[y:y+size, x:x+size]

profile = cv2.resize(square, (300, 300))

gray = cv2.cvtColor(profile, cv2.COLOR_BGR2GRAY)

cv2.imwrite("profile_color.jpg", profile)
cv2.imwrite("profile_gray.jpg", gray)

print("Colored Profile Picture:")
cv2_imshow(profile)

print("Black & White Profile Picture:")
cv2_imshow(gray)
```

```python
#Before & After Visualization
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread("passport_image.jfif")


original = img
img_50 = cv2.resize(img, None, fx=0.5, fy=0.5)
img_25 = cv2.resize(img, None, fx=0.25, fy=0.25)

height = 250

original = cv2.resize(
    original,
    (int(original.shape[1] * height / original.shape[0]), height)
)

img_50 = cv2.resize(
    img_50,
    (int(img_50.shape[1] * height / img_50.shape[0]), height)
)

img_25 = cv2.resize(
    img_25,
    (int(img_25.shape[1] * height / img_25.shape[0]), height)
)

# Combine
result = np.hstack((original, img_50, img_25))

cv2_imshow(result)
```

```python
#5.Dominant Color Extraction (HSV)
import cv2
import numpy as np

img = cv2.imread("passport_image.jfif")


hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

region = hsv[50:250, 100:300]

average = np.mean(region, axis=(0, 1))

H = average[0]
S = average[1]
V = average[2]

print("Average Hue (H):", H)
print("Average Saturation (S):", S)
print("Average Value (V):", V)

print("\nDominant Color Information:")
if H < 10 or H > 170:
    print("Red")
elif H < 35:
    print("Yellow / Orange")
elif H < 85:
    print("Green")
elif H < 130:
    print("Blue")
else:
    print("Purple")
```

```python
#Watermark Designer
import cv2
from google.colab.patches import cv2_imshow
img = cv2.imread("passport_image.jfif")


watermark = img.copy()


h, w = watermark.shape[:2]


cv2.rectangle(
    watermark,
    (0, h-70),
    (w, h),
    (0, 0, 0),
    -1
)


cv2.putText(
    watermark,
    "Soft Tech Institute",
    (30, h-25),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (255, 255, 255),
    2
)

cv2.imwrite("original.jpg", img)
cv2.imwrite("watermarked.jpg", watermark)

print("Original:")
cv2_imshow(img)

print("Watermarked:")
cv2_imshow(watermark)
```

```python
#Greeting Card Creator
import cv2
from google.colab.patches import cv2_imshow
img = cv2.imread("passport_image.jfif")

if img is None:
    print("Error: Could not find or load the image file.")
else:

    img = cv2.resize(img, (600, 400))

    cv2.putText(
        img,
        "Happy Birthday",
        (150, 210),
        cv2.FONT_HERSHEY_SIMPLEX,
        1.5,
        (255, 255, 255),
        3
    )
    cv2.circle(img, (80, 100), 30, (0, 0, 255), -1)
    cv2.circle(img, (520, 100), 30, (255, 0, 0), -1)
    cv2.circle(img, (300, 80), 30, (0, 255, 255), -1)

    cv2.line(img, (80, 130), (100, 300), (255, 255, 255), 2)
    cv2.line(img, (520, 130), (500, 300), (255, 255, 255), 2)
    cv2.line(img, (300, 110), (300, 300), (255, 255, 255), 2)
    cv2.line(img, (50, 350), (550, 350), (255, 255, 255), 3)
    cv2.imwrite("greeting_card.jpg", img)

    print("Greeting Card:")
    cv2_imshow(img)
```
